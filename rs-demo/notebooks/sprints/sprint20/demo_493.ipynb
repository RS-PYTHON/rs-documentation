### Filters can now be applied to /search endpoint, with operator predefined collections, to enrich query capabilities.  
In the example below, we use predefined collection adgs_aux_pp2 that returns only products with type AUX_PP2, and we select also products comming from "FOS" processing center.

```python
filter="processing:facility='FOS'"
print(f"There are {len(auxip_client.search(method='GET', collections = ['adgs_aux_pp2'], stac_filter=filter))} items by using the filter '{filter}'")
```

Request all products with type **AUX_PP2** sorted by eviction date. (Other ADGS sortables include: id (Name), auxip:id, file:size, type, created, start / end_datetime.


```python
filter="product:type='AUX_PP2'"
sortby = "-eviction_datetime"
print(f"There are {len(auxip_client.search(method='GET', stac_filter=filter, sortby=sortby))} items by using the filter '{filter}'")
```

CADIP sort parameters include: **id, start_datetime, datetime, end_datetime, published, platform, publicationDate, cadip:id, cadip:num_channels, cadip:station_unit_id, sat:absolute_orbit, cadip:acquisition_id, cadip:antenna_id, cadip:front_end_id, cadip:retransfer, cadip:antenna_status_ok, cadip:front_end_status_ok, cadip:planned_data_start, cadip:planned_data_stop, cadip:downlink_status_ok, cadip:delivery_push_ok**

```python
filter = "constellation='sentinel-2' AND cadip:retransfer='False'"
sortby = "-published"
pprint.pp(len(cadip_client.search(method="GET", stac_filter=filter, sortby=sortby)))
```
