# Read KeyCloak quota attribute

A new custom attribute is defined in Keycloak:

> **monthly-download-volume-limit-gb**: defines the maximum total volume of data in gigabytes that a user is allowed to download within any rolling 31-day period. 

This demo checks that the quota attribute is exposed by the API key and the OAuth2 token.

```python
from resources.utils import *
if not cluster_mode:
    print("NOTE: this demo only works in cluster mode, so we don't run it in local mode.")
    import sys
    sys.exit(0) # stop the papermill running

# Init the demos
init_demo()

# Reload the global vars again
from resources.utils import * 
```

```python
# Imports
import os
import requests
```

```python
# Get user info related to the api key
url = os.environ["RSPY_UAC_CHECK_URL"]
print(f"Call: {url!r}\n")

response = requests.get(url, headers={"x-api-key": os.environ["RSPY_APIKEY"]}).json()
assert "monthly-download-volume-limit-gb" in response["config"]
display(response)
```

```python
# Get user info related to the oauth2 token
url = f"{os.environ['RSPY_WEBSITE']}/auth/me"
print(f"Call: {url!r}\n")

session = requests.Session()
session.cookies.set("session", os.environ["RSPY_OAUTH2_COOKIE"])
response = session.get(url).json()
assert "monthly-download-volume-limit-gb" in response["attributes"]
display(response)
```

```python
# We have the same results when using the rs-client functions.
# Use any client: auxip or cadip or catalog or ...

print(f"Call 'RsClient.apikey_security()'\n")
response = auxip_client.apikey_security()
assert "monthly-download-volume-limit-gb" in response.attributes
display(response)

print(f"\nCall 'RsClient.apikey_attributes'\n")
response = auxip_client.apikey_attributes
assert "monthly-download-volume-limit-gb" in response
display(response)

print(f"\nCall 'RsClient.oauth2_security()'\n")
response = auxip_client.oauth2_security()
assert "monthly-download-volume-limit-gb" in response.attributes
display(response)

print(f"\nCall 'RsClient.oauth2_attributes'\n")
response = auxip_client.oauth2_attributes
assert "monthly-download-volume-limit-gb" in response
display(response)
```
