```python
# Do some init
import os
import json

rs_server_href = os.environ["RSPY_WEBSITE"]
```

```python
# Call the endpoint that synchronises RS users registered in Keycloak and the linked users from the Cloud provider (OVH)
# This endpoint is also a trigger for gathering the keycloak user attributes, roles and bucket configuration
response = http_session.post (f"{rs_server_href}/storage/accounts/update")
print(f"status_code = {response.status_code}")
print(response.content)
```

```python
import pprint
pp = pprint.PrettyPrinter(indent=4)
# Call the endpoint to get the OBS access rights of user osam
response = http_session.get (f"{rs_server_href}/storage/account/osam/rights")
assert response.status_code == 200
print(json.dumps(json.loads(response.content.decode()), indent=4))
```

```python
response = http_session.get(f"{rs_server_href}/storage/account/credentials")
print(json.dumps(json.loads(response.content.decode()), indent=4))
```
